```python
import requests
from bs4 import BeautifulSoup

all_books = []

for page in range(1, 6):
    url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for book in books:
        title = book.h3.a["title"]

        price_gbp = float(price_text.replace("Â£", "").replace("£", "").strip())

        rating_text = book.find("p", class_="star-rating")["class"][1]
        rating_map = {
            "One": 1,
            "Two": 2,
            "Three": 3,
            "Four": 4,
            "Five": 5
        }
        rating = rating_map.get(rating_text)

        availability = book.find(
            "p", class_="instock availability"
        ).get_text(" ", strip=True)

        in_stock = "In stock" in availability

        # Get the book's detail-page link
        book_link = book.h3.a["href"]
        detail_url = "https://books.toscrape.com/catalogue/" + book_link.replace("../", "")

        detail_response = requests.get(detail_url)
        detail_response.raise_for_status()

        detail_soup = BeautifulSoup(detail_response.text, "html.parser")

        breadcrumb = detail_soup.find("ul", class_="breadcrumb")
        breadcrumb_items = breadcrumb.find_all("li")

        category = breadcrumb_items[-2].get_text(strip=True)

        all_books.append({
            "Title": title,
            "price_gbp": price_gbp,
            "rating": rating,
            "availability": availability,
            "in_stock": in_stock,
            "category": category
        })

print("Total books:", len(all_books))
print("Total categories:", len(set(book["category"] for book in all_books)))
```

```python
import pandas as pd

# 1. Convert the 'all_books' list from your previous cell into a DataFrame
df = pd.DataFrame(all_books)

# 2. Standardize column name if needed
if 'Title' in df.columns:
    df.rename(columns={'Title': 'title'}, inplace=True)

# 3. Add the price_inr column using Masai's fixed baseline rate (1 GBP = 105.50 INR)
GBP_TO_INR = 105.50
df['price_inr'] = (df['price_gbp'] * GBP_TO_INR).round(2)

# 4. Convert string ratings to integers if they aren't already
rating_map = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}
if df['rating'].dtype == object:
    df['rating'] = df['rating'].map(rating_map)

# Display the prepared DataFrame
df.head()
```

```python
import sqlite3

# Connect to SQLite database file
conn = sqlite3.connect("data_pipeline.db")
cursor = conn.cursor()

# Enable Foreign Key constraints in SQLite
cursor.execute("PRAGMA foreign_keys = ON;")

# Reset tables if re-running
cursor.execute("DROP TABLE IF EXISTS books;")
cursor.execute("DROP TABLE IF EXISTS categories;")

# 1. Create normalized 2-table schema
cursor.execute("""
    CREATE TABLE categories (
        category_id INTEGER PRIMARY KEY AUTOINCREMENT,
        category_name TEXT UNIQUE NOT NULL
    );
""")

cursor.execute("""
    CREATE TABLE books (
        book_id INTEGER PRIMARY KEY AUTOINCREMENT,
        title TEXT NOT NULL,
        price_gbp REAL NOT NULL,
        price_inr REAL NOT NULL,
        rating INTEGER NOT NULL,
        in_stock INTEGER NOT NULL,
        category_id INTEGER NOT NULL,
        FOREIGN KEY (category_id) REFERENCES categories (category_id)
    );
""")

# 2. Insert unique categories into 'categories' table
unique_categories = df[['category']].drop_duplicates().rename(columns={'category': 'category_name'})
unique_categories.to_sql('categories', conn, if_exists='append', index=False)

# 3. Retrieve category IDs and map them to the books DataFrame
cat_lookup = pd.read_sql("SELECT category_id, category_name FROM categories;", conn)
df_merged = df.merge(cat_lookup, left_on='category', right_on='category_name')

# Ensure in_stock is 1 or 0 integer
df_merged['in_stock'] = df_merged['in_stock'].astype(int)

# 4. Insert books into 'books' table
books_to_insert = df_merged[['title', 'price_gbp', 'price_inr', 'rating', 'in_stock', 'category_id']]
books_to_insert.to_sql('books', conn, if_exists='append', index=False)

conn.close()
print("✓ SQLite Database successfully created and loaded with PK/FK constraints!")
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("data_pipeline.db")

queries = {
    "1. SELECT / WHERE / ORDER BY / LIMIT (5 Cheapest In-Stock Books)": """
        SELECT title, price_gbp, price_inr
        FROM books
        WHERE in_stock = 1
        ORDER BY price_gbp ASC
        LIMIT 5;
    """,

    "2. DISTINCT (Distinct Ratings Available)": """
        SELECT DISTINCT rating
        FROM books
        ORDER BY rating DESC;
    """,

    "3. IN (Books with 4 or 5 Star Ratings)": """
        SELECT title, rating, price_inr
        FROM books
        WHERE rating IN (4, 5)
        LIMIT 5;
    """,

    "4. BETWEEN (Books Priced Between 20.00 and 40.00 GBP)": """
        SELECT title, price_gbp, price_inr
        FROM books
        WHERE price_gbp BETWEEN 20.00 AND 40.00
        ORDER BY price_gbp DESC
        LIMIT 5;
    """,

    "5. JOIN (Top 10 Highest-Rated Books with Category Names)": """
        SELECT b.title, c.category_name, b.rating, b.price_gbp, b.price_inr
        FROM books b
        JOIN categories c ON b.category_id = c.category_id
        WHERE b.rating = 5
        ORDER BY b.price_gbp DESC
        LIMIT 10;
    """
}

for label, sql in queries.items():
    print(f"\n================ {label} ================")
    display(pd.read_sql(sql, conn))

conn.close()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("data_pipeline.db")

# Method A: Direct SQL JOIN
sql_join = """
    SELECT b.title, c.category_name, b.rating, b.price_gbp, b.price_inr
    FROM books b
    JOIN categories c ON b.category_id = c.category_id
    WHERE b.rating = 5
    ORDER BY b.price_gbp DESC
    LIMIT 10;
"""
sql_df = pd.read_sql(sql_join, conn)

# Method B: Pure Pandas read + pd.merge()
df_books = pd.read_sql("SELECT * FROM books;", conn)
df_categories = pd.read_sql("SELECT * FROM categories;", conn)
conn.close()

pandas_df = (
    pd.merge(df_books, df_categories, on="category_id")
    .query("rating == 5")
    .sort_values(by="price_gbp", ascending=False)
    [['title', 'category_name', 'rating', 'price_gbp', 'price_inr']]
    .head(10)
    .reset_index(drop=True)
)

sql_df = sql_df.reset_index(drop=True)

# Display both side by side
print("--- SQL JOIN Output ---")
display(sql_df)

print("\n--- Pandas pd.merge Output ---")
display(pandas_df)

# Equivalence Check Assert
assert sql_df.equals(pandas_df), "Mismatch between SQL and Pandas outputs!"
print("\n SUCCESS: Both SQL JOIN and Pandas pd.merge produced identical outputs!")
```
